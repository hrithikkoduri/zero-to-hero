```python
import random

# training split 
# build the dataset
def build_dataset(words): 
    block_size = 3
    X, Y = [], []

    for w in words:
        context = [0] * block_size
        for ch in w + '.':
            ix  = c2i[ch]
            X.append(context)
            # print(X)
            Y.append(ix)
            # print(Y)
            #print(''.join(i2c[i] for i in context), '------->', i2c[ix])
            context = context[1:] + [ix]
            # print(context)

    X = torch.tensor(X)
    Y = torch.tensor(Y)
    print(X.shape, Y.shape)
    return X, Y

random.seed = 42
random.shuffle(words)

n1 = int(0.8*len(words))
n2 = int(0.9*len(words))

print(len(words))
words_train, words_val, words_test = words[:n1], words[n1:n2], words[n2:]
X_train, Y_train = build_dataset(words_train)
X_val, Y_val = build_dataset(words_val)
X_test, Y_test = build_dataset(words_test)

print(X_train.shape, Y_train.shape)
print(X_val.shape, Y_val.shape)
print(X_test.shape, Y_test.shape)

```

## Sampling

```python
g = torch.Generator().manual_seed(2147483647)

for _ in range(20):
    out = []
    context = [0] * block_size
    while True:
        
        emb = C[torch.tensor([context])] # (1, 3, 10)
        h = torch.tanh(emb.view(1, -1) @ W1 + b1) # (1, 200)
        logits = h @ W2 + b2 # (1, 27)
        probs = F.softmax(logits, dim=1)
        ix = torch.multinomial(probs, num_samples=1, generator=g).item()
        context = context[1:] + [ix]
        out.append(ix)
        if ix == 0:
            break
    print(''.join(i2c[i] for i in out))
        
```
